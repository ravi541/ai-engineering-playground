```python
loader = UnstructuredFileLoader("/content/Python_inbuildfunctions.pdf")
```

```python
text_spliter = CharacterTextSplitter(chunk_size=2000,chunk_overlap=400)
```

```python
!pip install "unstructured[pdf]"
```

```python
document = loader.load()
texts = text_spliter.split_documents(document)
```

```python
texts[1]
```

```python
embeddings = HuggingFaceEmbeddings()
```

```python
persist_directory = "vector_db"
```

```python
vectordb = Chroma.from_documents(documents=texts,embedding=embeddings,persist_directory=persist_directory)
```

```python
retriever = vectordb.as_retriever()
```

```python
llm = ChatGroq(model = "llama-3.3-70b-versatile",temperature=1)
```

```python
qa_chain = RetrievalQA.from_chain_type(
    llm = llm,
    chain_type = "stuff",
    retriever = retriever,
    return_source_documents = True
)
```

```python
#invoke  the qa chain and get a respose from user query

query = "what does indian flag contains"
response = qa_chain.invoke({"query":query})
```

```python
print(response["result"])
```

```python
import gradio as gr

# Chat Function
def chatbot_response(question):

    try:
        response = qa_chain.invoke({"query": question})

        answer = response["result"]

        sources = []

        for doc in response["source_documents"]:
            if "source" in doc.metadata:
                sources.append(doc.metadata["source"])

        source_text = "\n".join(list(set(sources)))

        final_response = f"""
{answer}

-------------------
Source:
{source_text}
"""

        return final_response

    except Exception as e:
        return f"Error: {str(e)}"


# Gradio UI
demo = gr.Interface(
    fn=chatbot_response,
    inputs=gr.Textbox(
        lines=2,
        placeholder="Ask questions from your PDF..."
    ),
    outputs="text",
    title="RAG PDF Chatbot",
    description="LLMOps Project using LangChain + Groq + ChromaDB + Gradio",
    theme="soft"
)

demo.launch(debug=True)
```
