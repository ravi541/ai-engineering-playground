```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import load_img,img_to_array
```

```python
# Create an image data generator with augementations
datagen = ImageDataGenerator(
    rotation_range =40,
    width_shift_range=0.4,
    height_shift_range=0.4,
    shear_range=0.4,
    zoom_range=0.4,
    horizontal_flip=True,
    rescale=None,
    fill_mode='constant' #nearest,reflect,wrap,constant
)

#load image from path
img  = load_img(r"C:\Users\Raviteja\OneDrive\Desktop\CNN- Happy or Sad\training\sad\4images4.png")
img
```

```python
# convert to numpy array
x = img_to_array(img)
x = x.reshape((1,) +x.shape) # Reshape to (1,height,width,channels)
 
# Generate augemnted images and save

i = 0
for batch in datagen.flow(x,batch_size=1,
                          save_to_dir=r"C:\Users\Raviteja\AVSCODE\ai-engineering-playground\TENSOFLOW KERAS\DeepLearning\Data_Augmentation\DataAug_images",
                          save_prefix='man',save_format='jpeg'):
    i += 1
    if i > 5:
        break # prevent infinite loop
```
