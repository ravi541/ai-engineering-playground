```python
import numpy as np
import pandas as pd
import tensorflow as tf
```

```python
dataset= pd.read_csv(r"C:\Users\Raviteja\Downloads\Churn_Modelling.csv")
X=dataset.iloc[:,3:-1].values
y=dataset.iloc[:,-1].values
```

```python
dataset
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X[:,2]=le.fit_transform(X[:,2])
```

```python
X
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder',OneHotEncoder(),[1])],remainder='passthrough')
X=np.array(ct.fit_transform(X))
```

```python
from sklearn.preprocessing import StandardScaler
sc= StandardScaler()
X=sc.fit_transform(X)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test= train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

ANN MODEL BUILDING ARCHITECTURE

```python
ann = tf.keras.Sequential()  # establish layer one after another
```

```python
ann.add(tf.keras.layers.Dense(units=6)) # input layer with 6 neurons
```

```python
ann.add(tf.keras.layers.Dense(units=4,activation='relu')) #hidden layer
```

```python
ann.add(tf.keras.layers.Dense(units=1,activation='sigmoid')) #output layer
```

```python
ann.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
ann.fit(X_train,y_train,batch_size=32,epochs=5)
```

```python
y_pred = ann.predict(X_test)
y_pred = (y_pred>0.5)
print(np.concatenate((y_pred.reshape(len(y_pred),1),y_test.reshape(len(y_test),1)),1))
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_pred)
print(cm)
```

```python
from sklearn.metrics import accuracy_score
ac=accuracy_score(y_test,y_pred)
print(ac)
ac
```
